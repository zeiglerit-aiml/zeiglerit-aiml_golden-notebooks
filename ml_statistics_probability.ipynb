## Descriptive Statistics

Summarize the center, frequency, and spread of observed data.

### Mean

```python
def mean(values):
    if not values: raise ValueError("values cannot be empty")
    return sum(values) / len(values)

sample = [2, 4, 4, 6, 9]
print("Mean:", mean(sample))
```

**SDK visualization**

```python
x = np.array([2, 4, 4, 6, 9])
plt.figure(figsize=(6, 3))
plt.scatter(x, np.zeros_like(x), s=90, label="observations")
plt.axvline(x.mean(), color="crimson", label=f"mean = {x.mean():.2f}")
plt.yticks([]); plt.xlabel("Value"); plt.title("Mean as the balance point"); plt.legend(); plt.show()
```

### Median

```python
def median(values):
    if not values: raise ValueError("values cannot be empty")
    ordered = sorted(values)
    n = len(ordered); mid = n // 2
    return ordered[mid] if n % 2 else (ordered[mid-1] + ordered[mid]) / 2

sample = [2, 4, 4, 6, 15]
print("Median:", median(sample))
```

**SDK visualization**

```python
x = np.array([2, 4, 4, 6, 15])
plt.figure(figsize=(6, 3)); plt.scatter(x, np.zeros_like(x), s=90)
plt.axvline(np.median(x), color="crimson", label=f"median = {np.median(x):.1f}")
plt.yticks([]); plt.xlabel("Value"); plt.title("Median is the middle observation"); plt.legend(); plt.show()
```

### Mode

```python
from collections import Counter
def mode(values):
    if not values: raise ValueError("values cannot be empty")
    counts = Counter(values); highest = max(counts.values())
    return sorted([v for v, count in counts.items() if count == highest])

sample = [1, 2, 2, 3, 3, 4]
print("Mode(s):", mode(sample))
```

**SDK visualization**

```python
x = np.array([1, 2, 2, 3, 3, 4])
values, counts = np.unique(x, return_counts=True)
plt.figure(figsize=(6, 3)); plt.bar(values, counts)
plt.xlabel("Value"); plt.ylabel("Frequency"); plt.title("Mode: highest frequency"); plt.show()
```

### Variance

```python
def variance(values):
    if not values: raise ValueError("values cannot be empty")
    mu = sum(values) / len(values)
    return sum((x - mu)**2 for x in values) / len(values)

sample = [2, 4, 4, 6, 9]
print("Population variance:", variance(sample))
```

**SDK visualization**

```python
x = np.array([2, 4, 4, 6, 9]); mu = x.mean()
plt.figure(figsize=(6, 3)); plt.bar(range(len(x)), (x-mu)**2)
plt.axhline(np.var(x), color="crimson", label=f"variance = {np.var(x):.2f}")
plt.xlabel("Observation"); plt.ylabel("Squared deviation"); plt.title("Variance from squared deviations"); plt.legend(); plt.show()
```

### Standard Deviation

```python
def standard_deviation(values):
    return math.sqrt(variance(values))

sample = [2, 4, 4, 6, 9]
print("Population standard deviation:", standard_deviation(sample))
```

**SDK visualization**

```python
x = np.array([2, 4, 4, 6, 9]); mu, sigma = x.mean(), x.std()
plt.figure(figsize=(6, 3)); plt.hist(x, bins=np.arange(1, 11), alpha=.7, edgecolor="black")
plt.axvline(mu, color="crimson", label=f"mean = {mu:.2f}")
plt.axvspan(mu-sigma, mu+sigma, alpha=.2, label=f"±1 SD = {sigma:.2f}")
plt.xlabel("Value"); plt.ylabel("Count"); plt.title("Spread measured by standard deviation"); plt.legend(); plt.show()
```

## Probability & Distributions

Represent uncertainty, update beliefs with evidence, and describe data distributions.

### Probability

```python
def probability(favorable, total):
    if total <= 0 or favorable < 0 or favorable > total:
        raise ValueError("require 0 <= favorable <= total and total > 0")
    return favorable / total

print("P(rolling a 4 or higher on a fair die):", probability(3, 6))
```

**SDK visualization**

```python
outcomes = np.arange(1, 7); p = np.where(outcomes >= 4, 1/6, 0)
plt.figure(figsize=(6, 3)); plt.bar(outcomes, p)
plt.xlabel("Die outcome"); plt.ylabel("Probability contribution"); plt.title("Favorable outcomes: 4, 5, 6"); plt.show()
```

### Bayes’ Theorem

```python
def bayes(prior, likelihood, false_positive_rate):
    evidence = likelihood*prior + false_positive_rate*(1-prior)
    return likelihood*prior/evidence

print("Posterior:", bayes(prior=.01, likelihood=.95, false_positive_rate=.05))
```

**SDK visualization**

```python
prior, sensitivity, fpr = .01, .95, .05
posterior = bayes(prior, sensitivity, fpr)
labels = ["Prior P(A)", "Posterior P(A|B)"]
plt.figure(figsize=(6, 3)); plt.bar(labels, [prior, posterior])
plt.ylim(0, 1); plt.ylabel("Probability"); plt.title(f"Bayesian update: posterior = {posterior:.1%}"); plt.show()
```

### Normal Distribution

```python
def normal_pdf(x, mu=0.0, sigma=1.0):
    if sigma <= 0: raise ValueError("sigma must be positive")
    return math.exp(-0.5*((x-mu)/sigma)**2)/(sigma*math.sqrt(2*math.pi))

print("Standard normal density at 0:", normal_pdf(0))
```

**SDK visualization**

```python
x=np.linspace(-4,4,400); y=np.exp(-x**2/2)/math.sqrt(2*math.pi)
plt.figure(figsize=(6,3)); plt.plot(x,y); plt.fill_between(x,y,alpha=.2)
plt.xlabel("x"); plt.ylabel("Density"); plt.title("Standard normal distribution (μ=0, σ=1)"); plt.show()
```

## Relationships & Statistical Inference

Measure relationships and use sample data to evaluate claims about populations.

### Correlation

```python
def pearson_correlation(x, y):
    if len(x) != len(y) or len(x) < 2: raise ValueError("equal-length inputs; at least 2 values")
    mx, my = sum(x)/len(x), sum(y)/len(y)
    numerator = sum((a-mx)*(b-my) for a,b in zip(x,y))
    denominator = math.sqrt(sum((a-mx)**2 for a in x)*sum((b-my)**2 for b in y))
    if denominator == 0: raise ValueError("correlation undefined for constant input")
    return numerator/denominator

x=[1,2,3,4,5]; y=[2,3,5,4,7]
print("Pearson r:", pearson_correlation(x,y))
```

**SDK visualization**

```python
x=np.array([1,2,3,4,5]); y=np.array([2,3,5,4,7]); r=np.corrcoef(x,y)[0,1]
plt.figure(figsize=(5, 4)); plt.scatter(x,y); plt.title(f"Correlation (r = {r:.2f})")
plt.xlabel("X"); plt.ylabel("Y"); plt.grid(alpha=.25); plt.show()
```

### Regression

```python
def linear_regression(x, y):
    mx, my = sum(x)/len(x), sum(y)/len(y)
    slope = sum((a-mx)*(b-my) for a,b in zip(x,y))/sum((a-mx)**2 for a in x)
    return my-slope*mx, slope

x=[1,2,3,4,5]; y=[2,3,5,4,7]
print("intercept, slope:", linear_regression(x,y))
```

**SDK visualization**

```python
x=np.array([1,2,3,4,5]); y=np.array([2,3,5,4,7]); b0,b1=linear_regression(x.tolist(),y.tolist())
xx=np.linspace(x.min(),x.max(),100)
plt.figure(figsize=(5,4)); plt.scatter(x,y,label="observed"); plt.plot(xx,b0+b1*xx,color="crimson",label="least-squares fit")
plt.xlabel("X"); plt.ylabel("Y"); plt.title("Simple linear regression"); plt.legend(); plt.show()
```

### Hypothesis Testing

```python
def hypothesis_decision(p_value, alpha=0.05):
    if not 0 <= p_value <= 1: raise ValueError("p_value must be in [0, 1]")
    return "Reject H0" if p_value < alpha else "Fail to reject H0"

print(hypothesis_decision(p_value=.03, alpha=.05))
```

**SDK visualization**

### p-Value

```python
def two_sided_z_pvalue(z):
    return math.erfc(abs(z)/math.sqrt(2))

print("Two-sided p-value for z=2:", two_sided_z_pvalue(2))
```

**SDK visualization**

```python
z=np.linspace(-4,4,500); density=np.exp(-z**2/2)/math.sqrt(2*math.pi); cutoff=2
plt.figure(figsize=(6,3)); plt.plot(z,density); plt.fill_between(z,density,where=(z<=-cutoff)|(z>=cutoff),alpha=.35,label="two-sided p-value area")
plt.axvline(-cutoff,color="gray",ls="--"); plt.axvline(cutoff,color="gray",ls="--")
plt.xlabel("z statistic"); plt.ylabel("Density"); plt.title(f"Two-sided p-value = {two_sided_z_pvalue(2):.4f}"); plt.legend(); plt.show()
```

### Confidence Interval

```python
def mean_confidence_interval(values, confidence=0.95):
    n=len(values); m=sum(values)/n
    s=math.sqrt(sum((v-m)**2 for v in values)/(n-1))
    critical=stats.t.ppf((1+confidence)/2, df=n-1)
    margin=critical*s/math.sqrt(n)
    return m-margin, m+margin

sample=[12,14,15,13,16,14,15,17]
print("95% CI:", mean_confidence_interval(sample))
```

**SDK visualization**

```python
rng=np.random.default_rng(7); population=rng.normal(15,3, size=10000); intervals=[]
for _ in range(40): intervals.append(mean_confidence_interval(rng.choice(population, size=30).tolist()))
plt.figure(figsize=(6,4));
for i,(lo,hi) in enumerate(intervals): plt.plot([lo,hi],[i,i],color="steelblue")
plt.axvline(population.mean(),color="crimson",label="population mean"); plt.xlabel("Interval for sample mean"); plt.ylabel("Repeated sample"); plt.title("Illustrative 95% confidence intervals"); plt.legend(); plt.show()
```

### T-Test

```python
def one_sample_t_test(values, mu0=0):
    n=len(values); m=sum(values)/n
    s=math.sqrt(sum((v-m)**2 for v in values)/(n-1))
    t=(m-mu0)/(s/math.sqrt(n)); p=2*(1-stats.t.cdf(abs(t),df=n-1))
    return t,p

print("t statistic, two-sided p:", one_sample_t_test([5,6,7,8,9], mu0=6))
```

**SDK visualization**

```python
sample=np.array([5,6,7,8,9]); t,p=one_sample_t_test(sample.tolist(),6);
x=np.linspace(-4,4,400); y=stats.t.pdf(x,df=len(sample)-1)
plt.figure(figsize=(6,3)); plt.plot(x,y); plt.fill_between(x,y,where=np.abs(x)>=abs(t),alpha=.3,label="two-sided tail area")
plt.axvline(t,color="crimson",ls="--",label=f"t = {t:.2f}"); plt.xlabel("t statistic"); plt.title(f"One-sample t-test (p = {p:.4f})"); plt.legend(); plt.show()
```

### ANOVA

```python
def one_way_anova(*groups):
    k=len(groups); all_values=[v for g in groups for v in g]; n=len(all_values)
    grand=sum(all_values)/n
    ss_between=sum(len(g)*(sum(g)/len(g)-grand)**2 for g in groups)
    ss_within=sum(sum((v-sum(g)/len(g))**2 for v in g) for g in groups)
    dfb=k-1; dfw=n-k; f=(ss_between/dfb)/(ss_within/dfw)
    return f, stats.f.sf(f,dfb,dfw)

print("F statistic, p-value:", one_way_anova([4,5,6,5],[7,8,9,8],[5,6,5,7]))
```

**SDK visualization**

```python
groups=[np.array([4,5,6,5]),np.array([7,8,9,8]),np.array([5,6,5,7])]; f,p=one_way_anova(*[g.tolist() for g in groups])
plt.figure(figsize=(6,3)); plt.boxplot(groups,tick_labels=["A","B","C"]); plt.ylabel("Measurement"); plt.title(f"One-way ANOVA (F={f:.2f}, p={p:.4f})"); plt.show()
```

### Chi-Square Test

```python
def chi_square_gof(observed, expected):
    if len(observed)!=len(expected) or any(e<=0 for e in expected): raise ValueError("invalid expected counts")
    statistic=sum((o-e)**2/e for o,e in zip(observed,expected))
    return statistic, stats.chi2.sf(statistic,df=len(observed)-1)

print("chi-square, p-value:", chi_square_gof([18,22,27,33],[25,25,25,25]))
```

**SDK visualization**

```python
observed=np.array([18,22,27,33]); expected=np.repeat(observed.sum()/len(observed),len(observed))
plt.figure(figsize=(6,3)); idx=np.arange(4); plt.bar(idx-.18,observed,.36,label="Observed"); plt.bar(idx+.18,expected,.36,label="Expected")
plt.xticks(idx,["A","B","C","D"]); plt.ylabel("Count"); plt.title("Chi-square goodness-of-fit counts"); plt.legend(); plt.show()
```

### Decision Trees

```python
def gini_impurity(labels):
    n=len(labels)
    if n==0: return 0.0
    counts={label:labels.count(label) for label in set(labels)}
    return 1-sum((count/n)**2 for count in counts.values())

print("Gini impurity:", gini_impurity([0,0,0,1,1]))
```

### K-Means Clustering

```python
def assign_clusters(points, centroids):
    return [min(range(len(centroids)),key=lambda j:sum((a-b)**2 for a,b in zip(point,centroids[j]))) for point in points]

points=[[1,1],[1,2],[2,1],[8,8],[8,9],[9,8]]; centers=[[1,1],[8,8]]
print("Cluster assignments:", assign_clusters(points,centers))
```

**SDK visualization**

### Principal Component Analysis (PCA)

```python
def pca_2d(data):
    means=[sum(col)/len(col) for col in zip(*data)]
    centered=[[row[j]-means[j] for j in range(len(means))] for row in data]
    cov=np.cov(np.array(centered).T); eigenvalues,eigenvectors=np.linalg.eigh(cov)
    order=np.argsort(eigenvalues)[::-1]
    components=eigenvectors[:,order[:2]]
    return np.array(centered) @ components, eigenvalues[order]

projected,eigenvalues=pca_2d([[2,1],[3,2],[4,3],[5,4],[6,5]])
print("Projected coordinates:\n", projected)
```

**SDK visualization**

```python
rng=np.random.default_rng(12); X=rng.multivariate_normal([0,0],[ [3,2.4],[2.4,3] ],size=250)
Z, eigvals=pca_2d(X.tolist())
plt.figure(figsize=(5,4)); plt.scatter(Z[:,0],Z[:,1],s=18,alpha=.6); plt.axhline(0,color="gray",lw=.7); plt.axvline(0,color="gray",lw=.7)
plt.xlabel("Principal component 1"); plt.ylabel("Principal component 2"); plt.title("PCA projection"); plt.show()
```

### Time Series Forecasting

```python
def moving_average_forecast(values, window):
    if window<1 or len(values)<window: raise ValueError("window must be positive and <= data length")
    return sum(values[-window:])/window

series=[10,12,13,15,14,16,18]
print("Next-step 3-period forecast:", moving_average_forecast(series,3))
```

**SDK visualization**

```python
t=np.arange(1,25); series=10+0.4*t+2*np.sin(t/2)+np.random.default_rng(5).normal(0,.5,len(t))
window=3; forecasts=[np.mean(series[i-window:i]) for i in range(window,len(series))]
plt.figure(figsize=(7,3)); plt.plot(t,series,marker="o",label="observed"); plt.plot(t[window:],forecasts,ls="--",marker=".",label="3-period moving average")
plt.xlabel("Time"); plt.ylabel("Value"); plt.title("Time-series moving-average forecast"); plt.legend(); plt.show()
```

#### Geometric — (p)

Models the number of trials until the first success. Useful for modeling event counts and waiting processes.

#### Negative Binomial — (r, p)

Models the number of failures before a fixed number of successes. Can describe over-dispersed count data used in predictive modeling.

```python
# Pure Python — evaluate one value
import math
r, p, failures = 4, 0.45, 3
probability = math.comb(failures+r-1, r-1)*p**r*(1-p)**failures
print(f"P({failures} failures before {r} successes) = {probability:.5f}")
```

**SciPy + Matplotlib — visualize the same distribution.**

```python
d = stats.nbinom(n=4, p=0.45)
x = np.arange(0, 20)
plt.bar(x, d.pmf(x)); plt.title('Negative Binomial (r=4, p=0.45)')
plt.xlabel('Failures'); plt.ylabel('Probability'); plt.show()
```

#### Hypergeometric — (N, K, n)

Models sampling without replacement. Useful when observations are drawn from a finite population.

```python
# Pure Python — evaluate one value
import math
N, K, n, successes = 40, 12, 8, 3
probability = math.comb(K,successes)*math.comb(N-K,n-successes)/math.comb(N,n)
print(f"P({successes} successes in sample) = {probability:.5f}")
```

**SciPy + Matplotlib — visualize the same distribution.**

```python
d = stats.hypergeom(M=40, n=12, N=8)
x = np.arange(0, 9)
plt.bar(x, d.pmf(x)); plt.title('Hypergeometric (N=40, K=12, n=8)')
plt.xlabel('Successes in sample'); plt.ylabel('Probability'); plt.show()
```

#### Poisson — (λ)

Models event counts over a fixed interval. Often used for count prediction, arrivals, and incident-rate modeling.

```python
# Pure Python — evaluate one value
import math
lam, k = 4, 3
probability = math.exp(-lam)*lam**k/math.factorial(k)
print(f"P({k} events) = {probability:.5f}")
```

**SciPy + Matplotlib — visualize the same distribution.**

#### Binomial — (n, p)

### Continuous distributions

Positive values whose logarithms follow a Normal distribution. Common for skewed quantities such as transaction sizes or durations.

#### Softmax

Converts a vector of logits into class probabilities that sum to one; commonly used in multiclass output layers.

```python
import math

def softmax(values):
    shifted = [v-max(values) for v in values]
    exponentials = [math.exp(v) for v in shifted]
    return [v/sum(exponentials) for v in exponentials]

print('Softmax probabilities:', softmax([2.2, 0.4, -0.7]))
```

**NumPy SDK equivalent:**

```python
def softmax(x):
    e = np.exp(x - np.max(x))  # stable exponentials
    return e / e.sum()

logits = np.array([2.2, 0.4, -0.7])
print(softmax(logits), 'sum =', softmax(logits).sum())
```

### Loss functions — quantify prediction error

A loss function measures the difference between predictions and targets. During neural-network training, gradients of the loss guide parameter updates.

#### Mean Squared Error (MSE)

Squares residuals, penalizing large errors more strongly. Common for regression.

```python
def mse(y, pred):
    return np.mean((np.asarray(y) - np.asarray(pred))**2)

print(mse([1, 2, 3], [1.2, 1.8, 2.5]))
```

#### Mean Absolute Error (MAE)

Uses absolute residuals and is less sensitive to large outliers than MSE.

```python
def mae(y, pred):
    return np.mean(np.abs(np.asarray(y) - np.asarray(pred)))

print(mae([1, 2, 3], [1.2, 1.8, 2.5]))
```

#### Huber Loss

Quadratic for small errors and linear for large errors; useful when regression targets contain outliers.

```python
def huber(y, pred, delta=1.0):
    e = np.abs(np.asarray(y) - np.asarray(pred))
    return np.mean(np.where(e <= delta, 0.5*e**2, delta*(e - 0.5*delta)))

print(huber([1, 2, 3], [1.2, 1.8, 5.0]))
```

#### Binary Cross-Entropy (BCE)

Measures probability error for binary targets. Common with a sigmoid output for binary classification.

```python
def binary_cross_entropy(y, p, eps=1e-9):
    p = np.clip(p, eps, 1-eps)
    return -np.mean(y*np.log(p) + (1-y)*np.log(1-p))

print(binary_cross_entropy(np.array([0, 1, 1]), np.array([.1, .8, .6])))
```

#### Categorical Cross-Entropy

Measures probability assigned to the correct class. Common with softmax outputs in multiclass neural networks.

```python
def categorical_cross_entropy(y_onehot, p, eps=1e-9):
    return -np.mean(np.sum(y_onehot*np.log(np.clip(p, eps, 1)), axis=1))

y = np.array([[0,1,0], [1,0,0]])
p = np.array([[.1,.8,.1], [.7,.2,.1]])
print(categorical_cross_entropy(y, p))
```

#### Negative Log-Likelihood (NLL)

Penalizes low probability assigned to observed outcomes. A general likelihood-based objective used in probabilistic models.

```python
def nll(target_index, log_probs):
    return -np.mean(log_probs[np.arange(len(target_index)), target_index])

log_p = np.log([[.1,.8,.1], [.7,.2,.1]])
print(nll(np.array([1, 0]), log_p))
```

### Optimizers — update trainable parameters

Optimizers use gradients to change model parameters. The loss measures error; the optimizer decides how parameters move to reduce it.

#### Gradient Descent / SGD

Moves parameters opposite the gradient. The basic update rule used in neural-network training.

```python
w = 0.0
learning_rate = 0.1
gradient = 2*(w - 3)  # derivative of (w-3)^2
w -= learning_rate * gradient
print('Updated parameter:', w)
```

#### Momentum

Adds a running direction to reduce oscillation and accelerate progress along consistent gradients.

```python
w, velocity = 0.0, 0.0
learning_rate, beta = 0.1, 0.9
gradient = 2*(w - 3)
velocity = beta*velocity + gradient
w -= learning_rate*velocity
print('Updated parameter:', w)
```

#### RMSProp

Adapts each parameter's learning rate using a moving average of squared gradients; useful for differently scaled gradients.

```python
w, cache = 0.0, 0.0
learning_rate, beta, eps = 0.01, 0.9, 1e-8
g = 2*(w - 3)
cache = beta*cache + (1-beta)*g**2
w -= learning_rate*g/(np.sqrt(cache)+eps)
print('Updated parameter:', w)
```

#### Adam

Combines momentum-like first moments and adaptive second moments. Widely used for training deep neural networks and Transformers.

```python
w, m, v, t = 0.0, 0.0, 0.0, 1
lr, b1, b2, eps = 0.01, 0.9, 0.999, 1e-8
g = 2*(w-3)
m = b1*m + (1-b1)*g; v = b2*v + (1-b2)*g**2
mhat = m/(1-b1**t); vhat = v/(1-b2**t)
w -= lr*mhat/(np.sqrt(vhat)+eps)
print('Updated parameter:', w)
```

### Z-score standardization

\[
z = \frac{x-\mu}{\sigma}
\]

Here, \(x\) is an observation, \(\mu\) is the feature mean, and \(\sigma\) is its standard deviation. Use the training-set mean and standard deviation to transform validation and test data; do not fit scaling statistics on held-out data.

```python
# Pure Python and NumPy implementations
def z_score_python(values):
    mean = sum(values) / len(values)
    variance = sum((x - mean) ** 2 for x in values) / len(values)
    std = variance ** 0.5
    if std == 0:
        raise ValueError("Z-score is undefined for a constant feature")
    return [(x - mean) / std for x in values]

import numpy as np
def z_score_numpy(values):
    x = np.asarray(values, dtype=float)
    std = x.std(ddof=0)
    if std == 0:
        raise ValueError("Z-score is undefined for a constant feature")
    return (x - x.mean()) / std

sample = [10, 12, 14, 16, 18]
print("Python:", z_score_python(sample))
print("NumPy:", z_score_numpy(sample))
```

### Cosine similarity

\[
\operatorname{cosine}(A,B)=\frac{A\cdot B}{\|A\|_2\,\|B\|_2}
\]

The result measures orientation rather than magnitude. It is widely used for embedding/vector similarity. A zero vector has undefined cosine similarity, so the implementation rejects it.

```python
def cosine_similarity_python(a, b):
    if len(a) != len(b):
        raise ValueError("Vectors must have the same length")
    dot = sum(x*y for x, y in zip(a, b))
    na = sum(x*x for x in a) ** 0.5
    nb = sum(y*y for y in b) ** 0.5
    if na == 0 or nb == 0:
        raise ValueError("Cosine similarity is undefined for a zero vector")
    return dot / (na * nb)

def cosine_similarity_numpy(a, b):
    a, b = np.asarray(a, float), np.asarray(b, float)
    if a.ndim != 1 or b.ndim != 1 or a.shape != b.shape:
        raise ValueError("Inputs must be one-dimensional vectors of equal length")
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    if denom == 0:
        raise ValueError("Cosine similarity is undefined for a zero vector")
    return float(np.dot(a, b) / denom)

print(cosine_similarity_python([1, 2, 3], [2, 4, 6]))
print(cosine_similarity_numpy([1, 0], [0, 1]))
```

### Naive Bayes classification

\[
P(y\mid x_1,\ldots,x_n)\propto P(y)\prod_{i=1}^{n}P(x_i\mid y)
\]

Naive Bayes applies conditional independence between features given the class. This Gaussian Naive Bayes example estimates class priors, per-class feature means, and variances. It uses log probabilities to avoid numerical underflow and a variance floor for stability.

```python
import numpy as np

class GaussianNaiveBayes:
    def fit(self, X, y, var_smoothing=1e-9):
        X, y = np.asarray(X, float), np.asarray(y)
        self.classes_, counts = np.unique(y, return_counts=True)
        self.log_prior_ = np.log(counts / counts.sum())
        self.mean_, self.var_ = [], []
        floor = var_smoothing * max(float(X.var()), 1.0)
        for cls in self.classes_:
            rows = X[y == cls]
            self.mean_.append(rows.mean(axis=0))
            self.var_.append(np.maximum(rows.var(axis=0), floor))
        self.mean_, self.var_ = np.asarray(self.mean_), np.asarray(self.var_)
        return self

    def predict_log_proba(self, X):
        X = np.asarray(X, float)
        scores = []
        for prior, mean, var in zip(self.log_prior_, self.mean_, self.var_):
            log_likelihood = -0.5 * np.sum(
                np.log(2*np.pi*var) + (X - mean)**2 / var, axis=1
            )
            scores.append(prior + log_likelihood)
        scores = np.column_stack(scores)
        m = scores.max(axis=1, keepdims=True)
        return scores - (m + np.log(np.exp(scores-m).sum(axis=1, keepdims=True)))

    def predict(self, X):
        return self.classes_[self.predict_log_proba(X).argmax(axis=1)]

X = np.array([[1.0, 1.1], [1.2, 0.9], [4.0, 4.2], [3.8, 4.1]])
y = np.array(["A", "A", "B", "B"])
nb_model = GaussianNaiveBayes().fit(X, y)
print("Predictions:", nb_model.predict([[1.1, 1.0], [4.1, 4.0]]))
```

### Maximum likelihood estimation (MLE)

For independent observations from a normal distribution:

\[
\hat\mu_{MLE}=\frac{1}{n}\sum_{i=1}^{n}x_i,
\qquad
\hat\sigma^2_{MLE}=\frac{1}{n}\sum_{i=1}^{n}(x_i-\hat\mu)^2
\]

MLE chooses parameters that maximize the likelihood of the observed data. Note that the normal MLE variance uses denominator \(n\), not \(n-1\) (the latter is the usual unbiased sample-variance estimator).

```python
def normal_mle_python(values):
    mean = sum(values) / len(values)
    variance = sum((x - mean)**2 for x in values) / len(values)
    return mean, variance

def normal_mle_numpy(values):
    x = np.asarray(values, dtype=float)
    return float(x.mean()), float(x.var(ddof=0))

observations = [2.1, 1.8, 2.4, 2.0, 1.7, 2.2]
print("Python MLE (mu, variance):", normal_mle_python(observations))
print("NumPy MLE (mu, variance):", normal_mle_numpy(observations))
```

### Ordinary least squares (OLS) matrix solution

\[
\hat\beta=(X^TX)^{-1}X^Ty
\]

This closed form assumes the inverse exists. In practical numerical code, use a least-squares solver or pseudoinverse rather than explicitly calculating the inverse. Include a column of ones in \(X\) when fitting an intercept.

```python
def ols_python(X, y):
    # Normal equations with Gauss-Jordan elimination and partial pivoting.
    # Intended for small, well-conditioned teaching examples.
    X = [list(map(float, row)) for row in X]
    y = list(map(float, y))
    n_features = len(X[0])
    xtx = [[sum(row[i]*row[j] for row in X) for j in range(n_features)]
           for i in range(n_features)]
    xty = [sum(row[i]*target for row, target in zip(X, y))
           for i in range(n_features)]
    aug = [row + [rhs] for row, rhs in zip(xtx, xty)]
    for col in range(n_features):
        pivot = max(range(col, n_features), key=lambda r: abs(aug[r][col]))
        if abs(aug[pivot][col]) < 1e-12:
            raise ValueError("Singular design matrix; use a pseudoinverse/least-squares solver")
        aug[col], aug[pivot] = aug[pivot], aug[col]
        scale = aug[col][col]
        aug[col] = [v/scale for v in aug[col]]
        for row in range(n_features):
            if row != col:
                factor = aug[row][col]
                aug[row] = [a-factor*b for a, b in zip(aug[row], aug[col])]
    return [aug[i][-1] for i in range(n_features)]

def ols_numpy(X, y):
    X, y = np.asarray(X, float), np.asarray(y, float)
    return np.linalg.lstsq(X, y, rcond=None)[0]

x = [1., 2., 3., 4.]
X_design = [[1., value] for value in x]
y = [3., 5., 7., 9.]
beta_python = ols_python(X_design, y)
beta_numpy = ols_numpy(X_design, y)
print("Pure Python coefficients:", beta_python)
print("NumPy coefficients:", beta_numpy)
```

### MSE with L2 regularization (ridge objective)

\[
J(\beta)=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2
+\lambda\sum_{j=1}^{p}\beta_j^2
\]

The penalty discourages large coefficients. The example excludes the intercept from the penalty, a common convention. The value of \(\lambda\) should be selected using training/validation data, such as cross-validation.

```python
def mse_l2_python(y, y_pred, weights, lam):
    mse = sum((a-b)**2 for a, b in zip(y, y_pred)) / len(y)
    penalty = lam * sum(w*w for w in weights)  # pass non-intercept weights
    return mse + penalty

def mse_l2_numpy(y, y_pred, weights, lam):
    residual = np.asarray(y, float) - np.asarray(y_pred, float)
    w = np.asarray(weights, float)
    return float(np.mean(residual**2) + lam*np.sum(w**2))

target = [3., 5., 7.]
pred = [2.8, 5.1, 6.9]
print("Python objective:", mse_l2_python(target, pred, [1.9], 0.1))
print("NumPy objective:", mse_l2_numpy(target, pred, [1.9], 0.1))
```

### Eigenvalues and eigenvectors

\[
Av=\lambda v
\]

A nonzero vector \(v\) is an eigenvector of square matrix \(A\), and \(\lambda\) is its eigenvalue. Eigenvectors identify directions preserved by a linear transformation (up to scaling). For symmetric covariance matrices, eigenvectors are orthogonal and underpin PCA.

```python
A = np.array([[2.0, 1.0], [1.0, 2.0]])
eigenvalues, eigenvectors = np.linalg.eigh(A)  # symmetric matrix solver
for i, value in enumerate(eigenvalues):
    vector = eigenvectors[:, i]
    print(f"lambda={value:.3f}, Av={A @ vector}, lambda*v={value * vector}")
```

### Entropy and information gain

\[
H(Y)=-\sum_{i=1}^{k}p_i\log_2(p_i)
\]

Entropy measures uncertainty in a discrete distribution. Information gain is the reduction in entropy after splitting on a feature:

\[
IG(Y;X)=H(Y)-\sum_v P(X=v)H(Y\mid X=v)
\]

```python
import math

def entropy_python(probabilities):
    if any(p < 0 for p in probabilities) or not math.isclose(sum(probabilities), 1.0):
        raise ValueError("Probabilities must be nonnegative and sum to 1")
    return -sum(p * math.log2(p) for p in probabilities if p > 0)

def entropy_numpy(probabilities):
    p = np.asarray(probabilities, dtype=float)
    if p.ndim != 1 or np.any(p < 0) or not np.isclose(p.sum(), 1.0):
        raise ValueError("Probabilities must be a 1D nonnegative array summing to 1")
    p = p[p > 0]
    return float(-np.sum(p * np.log2(p)))

print("Balanced binary entropy:", entropy_python([0.5, 0.5]), "bits")
print("Skewed binary entropy:", entropy_numpy([0.9, 0.1]))
```

### Kullback–Leibler (KL) divergence

\[
D_{KL}(P\|Q)=\sum_x P(x)\log\frac{P(x)}{Q(x)}
\]

KL divergence measures how one probability distribution differs from another. It is nonnegative for valid distributions, is not symmetric, and is not a metric. The implementation assumes matching discrete supports and normalized distributions.

```python
def kl_divergence_python(p, q):
    if len(p) != len(q) or any(v < 0 for v in p) or any(v < 0 for v in q):
        raise ValueError("P and Q must have equal lengths and nonnegative values")
    if not np.isclose(sum(p), 1) or not np.isclose(sum(q), 1):
        raise ValueError("P and Q must each sum to 1")
    if any(qi <= 0 and pi > 0 for pi, qi in zip(p, q)):
        return float("inf")
    return sum(pi * math.log(pi/qi) for pi, qi in zip(p, q) if pi > 0)

def kl_divergence_numpy(p, q, eps=1e-12):
    p, q = np.asarray(p, float), np.asarray(q, float)
    if p.shape != q.shape or np.any(p < 0) or np.any(q < 0):
        raise ValueError("P and Q must be same-shaped nonnegative distributions")
    if not np.isclose(p.sum(), 1) or not np.isclose(q.sum(), 1):
        raise ValueError("P and Q must each sum to 1")
    mask = p > 0
    if np.any(q[mask] == 0):
        return float("inf")
    return float(np.sum(p[mask] * np.log(p[mask] / np.maximum(q[mask], eps))))

print("KL(P || Q):", kl_divergence_python([0.7, 0.3], [0.5, 0.5]), "nats")
```

### Support Vector Machine (SVM) hinge-loss objective

\[
\min_{w,b}\;\frac{1}{2}\|w\|_2^2
+C\sum_{i=1}^{n}\max\left(0,1-y_i(w^Tx_i+b)\right)
\]

Labels are encoded as \(y_i\in\{-1,+1\}\). The first term maximizes margin by controlling model complexity; hinge loss penalizes points inside the margin or on the wrong side. The following uses scikit-learn's linear SVM implementation.

### Singular Value Decomposition (SVD)

\[
A=U\Sigma V^T
\]

SVD factors a matrix into orthogonal left/right singular vectors and nonnegative singular values. It supports low-rank approximation, compression, denoising, and dimensionality reduction.

```python
A = np.array([[3., 1.], [1., 3.], [1., 1.]])
U, singular_values, VT = np.linalg.svd(A, full_matrices=False)
Sigma = np.diag(singular_values)
A_reconstructed = U @ Sigma @ VT
print("Singular values:", singular_values)
print("Reconstruction error:", np.linalg.norm(A - A_reconstructed))
# Rank-1 approximation
A_rank1 = U[:, :1] @ np.diag(singular_values[:1]) @ VT[:1, :]
print("Rank-1 approximation:\n", A_rank1)
```

### Lagrange multipliers

For an equality-constrained optimization problem:

\[
\max_x f(x)\quad\text{subject to }g(x)=0,
\qquad
\mathcal L(x,\lambda)=f(x)-\lambda g(x)
\]

At a regular constrained optimum, solve \(\nabla_x\mathcal L=0\) together with \(g(x)=0\). This small example maximizes \(xy\) subject to \(x+y=10\), whose optimum is \(x=y=5\).

```python
# Solve the stationarity equations:
# dL/dx = y - lambda = 0
# dL/dy = x - lambda = 0
# x + y = 10
# Therefore x = y = 5 and lambda = 5.
x_opt, y_opt, lambda_opt = 5.0, 5.0, 5.0
print("x, y, lambda:", x_opt, y_opt, lambda_opt)
print("Objective:", x_opt*y_opt, "| constraint residual:", x_opt+y_opt-10)
```

### 4.1 Proximal Policy Optimization (PPO)

PPO is an on-policy policy-gradient method. Its clipped surrogate objective limits the size of policy updates. This cell computes one clipped minibatch objective; it is not a full training loop.

```python
import torch
from torch.distributions import Categorical

logits = torch.tensor([[1.2, 0.1], [0.2, 0.8], [0.5, 0.4]], requires_grad=True)
actions = torch.tensor([0, 1, 0])
old_log_probs = torch.tensor([-0.35, -0.40, -0.55])
advantages = torch.tensor([1.0, 0.7, -0.3])
clip_eps = 0.2
new_log_probs = Categorical(logits=logits).log_prob(actions)
ratio = torch.exp(new_log_probs - old_log_probs)
loss = -torch.minimum(ratio * advantages,
                      torch.clamp(ratio, 1-clip_eps, 1+clip_eps) * advantages).mean()
loss.backward()
print(f"PPO clipped policy loss: {loss.item():.4f}; gradient norm: {logits.grad.norm().item():.4f}")
```

### 4.2 A* Search

A* prioritizes nodes by \(f(n)=g(n)+h(n)\): path cost so far plus estimated remaining cost. With nonnegative edges and an admissible heuristic, it can find an optimal path.

```python
import heapq

graph = {"A":{"B":1,"C":4}, "B":{"C":2,"D":5}, "C":{"D":1}, "D":{}}
heuristic = {"A":3,"B":2,"C":1,"D":0}
def astar(graph, start, goal, h):
    frontier=[(h[start],0,start,[start])]; best={start:0}
    while frontier:
        _, cost, node, path=heapq.heappop(frontier)
        if node==goal: return path,cost
        if cost != best[node]: continue
        for nxt, edge in graph[node].items():
            candidate=cost+edge
            if candidate < best.get(nxt,float("inf")):
                best[nxt]=candidate
                heapq.heappush(frontier,(candidate+h[nxt],candidate,nxt,path+[nxt]))
    return None,float("inf")
print("A* path and cost:", astar(graph,"A","D",heuristic))
```

### 4.3 Monte Carlo Tree Search (MCTS)

MCTS alternates selection, expansion, rollout, and reward backpropagation. This fixed-action toy illustrates UCB1 selection and visit counts; real game agents need state transitions, terminal tests, and rollout policies.

```python
import math, random
random.seed(7)
class Candidate:
    def __init__(self, reward): self.reward,self.visits=reward,0
candidates=[Candidate(.2),Candidate(.8),Candidate(.5)]
for _ in range(500):
    total=sum(c.visits for c in candidates)
    child=max(candidates,key=lambda c:c.reward+math.sqrt(2*math.log(total+1)/(c.visits+1)))
    child.visits+=1
print("(expected reward, visits):",[(c.reward,c.visits) for c in candidates])
```

### 4.6 Alpha-Beta Pruning

Alpha-beta pruning accelerates minimax by skipping branches that cannot change the final decision. It returns the same minimax value for a fixed tree and move ordering.

```python
def alphabeta(tree,node,depth,alpha,beta,maximizing):
    if depth==0 or isinstance(tree[node],(int,float)): return tree[node]
    if maximizing:
        value=float("-inf")
        for child in tree[node]:
            value=max(value,alphabeta(tree,child,depth-1,alpha,beta,False)); alpha=max(alpha,value)
            if alpha>=beta: break
        return value
    value=float("inf")
    for child in tree[node]:
        value=min(value,alphabeta(tree,child,depth-1,alpha,beta,True)); beta=min(beta,value)
        if beta<=alpha: break
    return value
tree={"root":["left","right"],"left":["a","b"],"right":["c","d"],"a":3,"b":5,"c":2,"d":9}
print("Minimax value:",alphabeta(tree,"root",2,float("-inf"),float("inf"),True))
```

### 4.8 Generative Adversarial Network (GAN)

A GAN trains a generator to create samples and a discriminator to distinguish generated from real data. This small one-dimensional Gaussian demo is for understanding alternating updates, not a stable image-generation recipe.

```python
import torch
from torch import nn

torch.manual_seed(3)
G=nn.Sequential(nn.Linear(4,16),nn.ReLU(),nn.Linear(16,1))
D=nn.Sequential(nn.Linear(1,16),nn.LeakyReLU(.2),nn.Linear(16,1),nn.Sigmoid())
opt_g=torch.optim.Adam(G.parameters(),lr=1e-3); opt_d=torch.optim.Adam(D.parameters(),lr=1e-3); bce=nn.BCELoss()
for _ in range(100):
    real=torch.randn(32,1)*.7+2; fake=G(torch.randn(32,4)).detach()
    dloss=bce(D(real),torch.ones(32,1))+bce(D(fake),torch.zeros(32,1))
    opt_d.zero_grad(); dloss.backward(); opt_d.step()
    generated=G(torch.randn(32,4)); gloss=bce(D(generated),torch.ones(32,1))
    opt_g.zero_grad(); gloss.backward(); opt_g.step()
print("Generated sample mean:",G(torch.randn(256,4)).mean().item())
```

### 4.10 Behavior Trees

Behavior trees compose conditions and actions. A sequence stops at the first non-success result; this simple version illustrates bounded decision flow for robotics, games, and agent workflows.

```python
SUCCESS, FAILURE, RUNNING = "SUCCESS", "FAILURE", "RUNNING"
class Sequence:
    def __init__(self,*children): self.children=children
    def tick(self,context):
        for child in self.children:
            result=child(context)
            if result!=SUCCESS: return result
        return SUCCESS
def has_target(ctx): return SUCCESS if ctx.get("target_visible") else FAILURE
def within_range(ctx): return SUCCESS if ctx.get("distance",999)<5 else FAILURE
def engage(ctx): ctx["action"]="engage_target"; return SUCCESS
world={"target_visible":True,"distance":3.2}
print("Behavior tree:",Sequence(has_target,within_range,engage).tick(world),world)
```

### 4.11 Soft Actor-Critic (SAC)

SAC is an off-policy actor-critic method for continuous control. It maximizes expected reward plus entropy. This cell computes a toy entropy-regularized Gaussian-policy objective; a complete SAC agent also needs twin Q-networks, target networks, and a replay buffer.

```python
import torch
from torch.distributions import Normal

mean=torch.tensor([.2,-.1],requires_grad=True); log_std=torch.tensor([-.5,-.3],requires_grad=True)
dist=Normal(mean,log_std.exp()); action=dist.rsample(); log_prob=dist.log_prob(action).sum()
estimated_q=-(action-.7).square().sum(); alpha=.2
actor_loss=alpha*log_prob-estimated_q; actor_loss.backward()
print("SAC-style actor loss:",actor_loss.item(),"policy gradient norm:",torch.stack([mean.grad.norm(),log_std.grad.norm()]).norm().item())
```
